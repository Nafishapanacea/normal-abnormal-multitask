```python
import numpy as np
import cv2
import os
from tqdm import tqdm
from PIL import Image
```

```python
def mild_clahe_pil(pil_img):
    gray = np.array(pil_img.convert("L"))

    clahe = cv2.createCLAHE(
        clipLimit=1.2,
        tileGridSize=(16, 16)
    )

    enhanced = clahe.apply(gray)

    enhanced_rgb = np.stack([enhanced]*3, axis=-1)

    return Image.fromarray(enhanced_rgb)
```

```python
def process_folder(in_dir, out_dir):
    os.makedirs(out_dir, exist_ok=True)

    for fname in tqdm(os.listdir(in_dir)):
        if not fname.lower().endswith((".png", ".jpg", ".jpeg")):
            continue

        img_path = os.path.join(in_dir, fname)
        out_path = os.path.join(out_dir, fname)
        print(img_path , out_path)

        img = Image.open(img_path).convert("RGB")
        img = mild_clahe_pil(img)
        img.save(out_path)

# # Example
process_folder(
    in_dir="/home/ubuntu/Documents/Nafisha/chest-xray-NormalAbnormal/shenzhen/images/",
    out_dir="/home/ubuntu/Documents/Nafisha/chest-xray-NormalAbnormal/shenzhen_CLAHE"
)
```
